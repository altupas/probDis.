Item 1

```r
mean_signal <- 200
sd_signal <- sqrt(256)

p_exceed_224 <- pnorm(224, mean=mean_signal, sd=sd_signal, lower.tail=FALSE)
p_exceed_224

p_between_186_224 <- pnorm(224, mean=mean_signal, sd=sd_signal) - pnorm(186, mean=mean_signal, sd=sd_signal)
p_between_186_224

voltage_25_percent <- qnorm(0.25, mean=mean_signal, sd=sd_signal)
voltage_25_percent

numerator <- pnorm(240, mean=mean_signal, sd=sd_signal) - pnorm(210, mean=mean_signal, sd=sd_signal)
denominator <- pnorm(210, mean=mean_signal, sd=sd_signal, lower.tail=FALSE)
p_given_210 <- numerator / denominator
p_given_210

q3 <- qnorm(0.75, mean=mean_signal, sd=sd_signal)
q1 <- qnorm(0.25, mean=mean_signal, sd=sd_signal)
iqr <- q3 - q1
iqr

numerator2 <- pnorm(220, mean=mean_signal, sd=sd_signal) - pnorm(210, mean=mean_signal, sd=sd_signal)
denominator2 <- pnorm(210, mean=mean_signal, sd=sd_signal, lower.tail=FALSE)
p_less220_given_greater210 <- numerator2 / denominator2
p_less220_given_greater210

numerator3 <- pnorm(220, mean=mean_signal, sd=sd_signal, lower.tail=FALSE)
denominator3 <- pnorm(200, mean=mean_signal, sd=sd_signal, lower.tail=FALSE)
p_greater220_given_greater200 <- numerator3 / denominator3
p_greater220_given_greater200
```

```r
mean_downtime <- 25
sd_downtime <- sqrt(144)

lower_bound_95 <- qnorm(0.025, mean=mean_downtime, sd=sd_downtime)
upper_bound_95 <- qnorm(0.975, mean=mean_downtime, sd=sd_downtime)
bounds_95 <- c(lower_bound_95, upper_bound_95)
bounds_95

bound_90th_percentile <- qnorm(0.90, mean=mean_downtime, sd=sd_downtime)
bound_90th_percentile
```
